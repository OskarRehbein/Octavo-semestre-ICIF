## ¿Qué es NumPy?

[NumPy](https://numpy.org/doc/stable/index.html) significa Python numérico (numerical Python). Es la columna vertebral de todo tipo de cálculo científico y numérico en Python.

## ¿Por qué NumPy?

Puedes hacer cálculos numéricos usando Python puro. Al principio puede parecerte que Python es rápido, pero en cuanto tus datos crecen empiezas a notar la lentitud.

Una de las razones principales para usar NumPy es que es rápido. Tras bambalinas, el código está optimizado para ejecutarse en C, que es otro lenguaje de programación capaz de hacer cosas mucho más rápido que Python.

La ventaja de que esto ocurra tras bambalinas es que no necesitas saber nada de C para aprovecharlo. Puedes escribir tus cálculos numéricos en Python usando NumPy y obtener igualmente el beneficio de velocidad.

Si tienes curiosidad sobre qué produce esa mejora de velocidad, se trata de un proceso llamado vectorización. La [vectorización](https://en.wikipedia.org/wiki/Vectorization) busca hacer los cálculos evitando bucles, ya que los bucles pueden convertirse en cuellos de botella.

NumPy logra la vectorización mediante un proceso llamado [broadcasting](https://docs.scipy.org/doc/numpy/user/basics.broadcasting.html#module-numpy.doc.broadcasting).

NumPy usa números pseudoaleatorios, lo que significa que los números parecen aleatorios pero en realidad no lo son: están predeterminados.

Por consistencia, puede que quieras que los números aleatorios que generas sean los mismos a lo largo de tus experimentos.

Para hacerlo, puedes usar [`np.random.seed()`](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.seed.html).

Lo que hace es decirle a NumPy: "Oye, quiero que crees números aleatorios pero mantenlos alineados con la semilla".

Veámoslo.

```python
# Fijar la semilla aleatoria en 0
np.random.seed(0)

# Generar números 'aleatorios'
np.random.randint(10, size=(5, 3))
```

Con `np.random.seed()` fijado, cada vez que ejecutes la celda de arriba se generarán los mismos números aleatorios.

¿Y si `np.random.seed()` no estuviera fijado?

Cada vez que ejecutes la celda de abajo aparecerá un conjunto de números distinto.

```python
# Generar más números aleatorios
np.random.randint(10, size=(5, 3))
```

Veámoslo en acción otra vez; mantendremos la consistencia fijando la semilla aleatoria en 0.

```python
# Fijar la semilla aleatoria en el mismo número que arriba
np.random.seed(0)

# Salen los mismos números aleatorios
np.random.randint(10, size=(5, 3))
```

Como `np.random.seed()` está fijado en 0, los números aleatorios son los mismos que los de la celda que también tenía `np.random.seed()` fijado en 0.

Fijar `np.random.seed()` no es 100% necesario, pero ayuda a mantener los mismos números a lo largo de tus experimentos.

Por ejemplo, supongamos que quieres dividir tus datos aleatoriamente en conjuntos de entrenamiento y prueba.

Cada vez que dividas al azar, podrías obtener filas distintas en cada conjunto.

Si compartieras tu trabajo con otra persona, esa persona también obtendría filas distintas en cada conjunto.

Fijar `np.random.seed()` garantiza que siga habiendo aleatoriedad, solo que hace que esa aleatoriedad sea repetible. De ahí el nombre de números 'pseudoaleatorios'.

```python
np.random.seed(0)
df = pd.DataFrame(np.random.randint(10, size=(5, 3)))
df
```

### Broadcasting

- ¿Qué es el broadcasting?
    - El broadcasting es una característica de NumPy que realiza una operación a lo largo de múltiples dimensiones de los datos sin replicarlos. Esto ahorra tiempo y espacio. Por ejemplo, si tienes un arreglo de 3x3 (A) y quieres sumarle un arreglo de 1x3 (B), NumPy sumará la fila de (B) a cada fila de (A).

- Reglas del broadcasting
    1. Si los dos arreglos difieren en su número de dimensiones, la forma del que tiene menos dimensiones se rellena con unos por su lado inicial (izquierdo).
    2. Si la forma de los dos arreglos no coincide en alguna dimensión, el arreglo cuya forma es igual a 1 en esa dimensión se estira para coincidir con la otra forma.
    3. Si en alguna dimensión los tamaños no coinciden y ninguno es igual a 1, se lanza un error.


**La regla del broadcasting:**
Para poder hacer broadcasting, el tamaño de los ejes finales de ambos arreglos de una operación debe ser el mismo, o bien uno de ellos debe ser uno.

**¿Qué es la media?**

La media es lo mismo que el promedio. Puedes encontrar el promedio de un conjunto de números sumándolos y dividiéndolos por la cantidad de números que hay.

**¿Qué es la desviación estándar?**

La [desviación estándar](https://www.mathsisfun.com/data/standard-deviation.html) es una medida de qué tan dispersos están los números.

**¿Qué es la varianza?**

La [varianza](https://www.mathsisfun.com/data/standard-deviation.html) es el promedio de las diferencias al cuadrado respecto de la media.

Para calcularla:
1. Calcula la media
2. Para cada número, resta la media y eleva el resultado al cuadrado
3. Calcula el promedio de esas diferencias al cuadrado

```python
# Demostración de la varianza
high_var_array = np.array([1, 100, 200, 300, 4000, 5000])
low_var_array = np.array([2, 4, 6, 8, 10])

np.var(high_var_array), np.var(low_var_array)
```

```python
np.std(high_var_array), np.std(low_var_array)
```

```python
# La desviación estándar es la raíz cuadrada de la varianza
np.sqrt(np.var(high_var_array))
```

### Producto punto

Las dos reglas principales del producto punto que hay que recordar son:

1. Las **dimensiones internas** deben coincidir:
  * `(3, 2) @ (3, 2)` no funciona
  * `(2, 3) @ (3, 2)` sí funciona
  * `(3, 2) @ (2, 3)` sí funciona

2. La matriz resultante tiene la forma de las **dimensiones externas**:
 * `(2, 3) @ (3, 2)` -> `(2, 2)`
 * `(3, 2) @ (2, 3)` -> `(3, 3)`

**Nota:** En NumPy, `np.dot()` y `@` se pueden usar para obtener el mismo resultado en arreglos de 1 a 2 dimensiones. Sin embargo, su comportamiento empieza a diferir en arreglos de 3 o más dimensiones.

```python
np.random.seed(0)
mat1 = np.random.randint(10, size=(3, 3))
mat2 = np.random.randint(10, size=(3, 2))

mat1.shape, mat2.shape
```

```python
mat1
```

```python
mat2
```

```python
np.dot(mat1, mat2)
```

```python
# También se puede lograr np.dot() con "@" 
# (sin embargo, pueden comportarse distinto en arreglos de 3D o más)
mat1 @ mat2
```

```python
np.random.seed(0)
mat3 = np.random.randint(10, size=(4,3))
mat4 = np.random.randint(10, size=(4,3))
mat3
```

```python
mat4
```

```python
# Esto fallará porque las dimensiones internas de las matrices no coinciden
np.dot(mat3, mat4)
```

```python
# Producto punto
np.dot(mat3.T, mat4)
```

```python
# Multiplicación elemento a elemento, también conocida como producto de Hadamard
mat3 * mat4
```

### Ejemplo práctico de producto punto: ventas de mantequillas de frutos secos

```python
np.random.seed(0)
sales_amounts = np.random.randint(20, size=(5, 3))
sales_amounts
```

```python
weekly_sales = pd.DataFrame(sales_amounts,
                            index=["Mon", "Tues", "Wed", "Thurs", "Fri"],
                            columns=["Almond butter", "Peanut butter", "Cashew butter"])
weekly_sales
```

```python
prices = np.array([10, 8, 12])
prices
```

```python
butter_prices = pd.DataFrame(prices.reshape(1, 3),
                             index=["Price"],
                             columns=["Almond butter", "Peanut butter", "Cashew butter"])
butter_prices.shape
```

```python
weekly_sales.shape
```

```python
# Calcular el total de ventas de un día completo
total_sales = prices.dot(sales_amounts)
total_sales
```

Las formas no están alineadas: necesitamos que los dos números del medio sean iguales.

```python
prices
```

```python
sales_amounts.T.shape
```

```python
# Para que los números del medio coincidan, podemos transponer
total_sales = prices.dot(sales_amounts.T)
total_sales
```

```python
butter_prices.shape, weekly_sales.shape
```

```python
daily_sales = butter_prices.dot(weekly_sales.T)
daily_sales
```

```python
# Hay que transponer de nuevo
weekly_sales["Total"] = daily_sales.T
weekly_sales
```

* [`np.argmin()`](https://numpy.org/doc/stable/reference/generated/numpy.argmin.html) - devuelve el índice o los índices que dan el o los valores más bajos a lo largo de un eje.
